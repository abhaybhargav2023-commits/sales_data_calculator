# Sales Data Calculator

This project analyzes a collection of manually created sales values using Python.

The program calculates:
- Total sales
- Average sales
- Highest sale
- Lowest sale

Python's built-in functions are used for the calculations.

```python

sales = [12000, 15000, 9000, 18000, 14000]

print("Sales Data:", sales)
```

```python

total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
# Final Sales Analysis

print("===== Sales Data Analysis =====")
print(f"Number of Sales: {len(sales)}")
print(f"Total Sales: ₹{total_sales}")
print(f"Average Sales: ₹{average_sales:.2f}")
print(f"Highest Sale: ₹{highest_sale}")
print(f"Lowest Sale: ₹{lowest_sale}")
```
